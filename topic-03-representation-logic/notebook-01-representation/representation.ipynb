## Challenge 3: equivalent expressions

Predict all four rows for De Morgan's first law. Add columns for intermediate operations in your paper trace.

```python
for a in [False, True]:
    for b in [False, True]:
        left = not (a and b)
        right = (not a) or (not b)
        print(int(a), int(b), int(left), int(right))
        assert left == right
```
